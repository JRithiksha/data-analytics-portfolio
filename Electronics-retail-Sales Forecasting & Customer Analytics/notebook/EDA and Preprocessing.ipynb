```python
# ==============================
# EDA + PREPROCESSING FILE
# ==============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pickle

# Load dataset
df = pd.read_csv("C:\\Users\\HP\\Downloads\\ML project\\Electronic-Retail-Sales-Analysis-ML\\data\\Electronic_sales_Sep2023-Sep2024.csv")

# ------------------------------
# BASIC UNDERSTANDING
# ------------------------------
print("Dataset Shape:", df.shape)
print(df.info())
print(df.head())

# ------------------------------
# HANDLE MISSING VALUES
# ------------------------------
df.fillna(method='ffill', inplace=True)

# ------------------------------
# FEATURE ENGINEERING
# ------------------------------
# Convert Date column
df['Date'] = pd.to_datetime(df['Date'])

df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day

# ------------------------------
# CREATE TARGET COLUMN
# ------------------------------
if 'Sales' not in df.columns:
    df['Sales'] = df['Price'] * df['Quantity']

# ------------------------------
# EDA VISUALIZATIONS
# ------------------------------

# 1. Sales Trend
plt.figure(figsize=(10,5))
df.groupby('Date')['Sales'].sum().plot()
plt.title("Sales Trend Over Time")
plt.savefig("sales_trend.png")
plt.close()

# 2. Top Products
top_products = df.groupby('Product')['Sales'].sum().sort_values(ascending=False).head(10)
print("\nTop Products:\n", top_products)

# 3. Correlation Heatmap
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.savefig("correlation.png")
plt.close()

# ------------------------------
# ENCODING
# ------------------------------
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Product'] = le.fit_transform(df['Product'])

# Save encoder
pickle.dump(le, open("label_encoder.pkl", "wb"))

# ------------------------------
# SAVE CLEANED DATA
# ------------------------------
df.to_csv("cleaned_data.csv", index=False)

print("\n✅ EDA & Preprocessing Completed. Cleaned data saved.")
```
